### **Data Manipulation using Pandas**

*Importing the Data*

```python
import pandas as pd
df = pd.read_csv('/ai_job_market_insights.csv')
df
```

*Columns and thier Data Types*

```python
df.dtypes
```

*Number of Rows*

```python
print(df.iloc[:,:].count)
```

*Making sure the data doesn't countain any null or empty values // Data Cleansing*
looks like there are no Nan or Null values in the data.

```python
df_cleaned = df.dropna(subset=['Job_Title', 'Industry', 'Company_Size', 'Location',
                               'AI_Adoption_Level', 'Automation_Risk', 'Required_Skills',
                               'Salary_USD', 'Remote_Friendly', 'Job_Growth_Projection'])
print(df_cleaned)
print(df_cleaned.iloc[:,:].count)
```

Observation 1: Get all the rows where the better AI adaptation level is positively affecting the job growth prediction

```python
df_new_table1 = df.loc[(df['AI_Adoption_Level'].isin(['High','Medium'])) & (df['Job_Growth_Projection'].isin(['Growth','Stable']))]
df_new_table1 = df_new_table1.sort_values(['AI_Adoption_Level','Job_Growth_Projection'])
df_new_table1
```

Observation 2: Get all the rows based on indsutry and the Job titles involved with the industry and the skills required for each job

```python
df_new_table2 = df.copy()
df_new_table2['Group of Skills required'] = df.groupby(['Job_Title','Industry'])['Required_Skills'].transform(lambda x: ', '.join(x))
df_new_table2 = df_new_table2[['Job_Title','Industry','Group of Skills required']]
df_new_table2.drop_duplicates(inplace=True)
df_new_table2
```

How does company size impact the adoption of remote work and salaries?

```python
df_new_table3 = df.groupby(['Company_Size'])[['Salary_USD', 'Remote_Friendly']].agg({'Salary_USD': 'mean', 'Remote_Friendly': 'count'})
df_new_table3 = df_new_table3.rename(columns={'Remote_Friendly': 'Count of Remote Firendly jobs'})
df_new_table3
```

```python
df_new_table4 = df.groupby(['AI_Adoption_Level', 'Company_Size'])['Job_Growth_Projection'].transform('count')

df_new_table4
```

```python
df_new_table5 = df[['Job_Title','Industry','Remote_Friendly', 'Salary_USD']]
df_new_table5 = df_new_table5.loc[df['Remote_Friendly'] == 'Yes']
df_new_table5 = df_new_table5.groupby(['Job_Title', 'Industry'])['Salary_USD'].mean()
df_new_table5
```
